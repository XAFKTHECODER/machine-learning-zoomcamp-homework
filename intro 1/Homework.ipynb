```python
import pandas as pd
import numpy as np
```

```python
print(f"Q1 - Version de Pandas : {pd.__version__}")
```

```python
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"
df = pd.read_csv(url)
```

```python
num_records = len(df)
print(f"Q2 - Nombre de lignes : {num_records}")
```

```python
num_fuel_types = df['fuel_type'].nunique()
fuel_types_list = df['fuel_type'].unique().tolist()
print(f"Q3 - Nombre de types de carburants : {num_fuel_types} ({fuel_types_list})")
```

```python
missing_series = df.isnull().sum()
num_cols_missing = (missing_series > 0).sum()
print(f"Q4 - Nombre de colonnes avec valeurs manquantes : {num_cols_missing}")
print("     Détails :")
print(missing_series[missing_series > 0])
```

```python
asia_cars = df[df['origin'] == 'Asia']
max_asia_efficiency = asia_cars['fuel_efficiency_mpg'].max()
print(f"Q5 - Efficacité maximale (MPG) pour les véhicules d'Asie : {max_asia_efficiency}")
```

```python
median_before = df['horsepower'].median()
```

```python
mode_hp = df['horsepower'].mode()[0]
df_filled = df.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(mode_hp)
median_after = df_filled['horsepower'].median()
```

```python
print(f"Q6 - Médiane initiale : {median_before}")
print(f"     Mode utilisé (fillna) : {mode_hp}")
print(f"     Médiane après remplacement : {median_after}")

if median_after < median_before:
    print("     Résultat : Yes, it decreased")
elif median_after > median_before:
    print("     Résultat : Yes, it increased")
else:
    print("     Résultat : No")
```

```python
asia_subset = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']]
```

```python
asia_7 = asia_subset.head(7)
```

```python
X = asia_7.values
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = (XTX_inv.dot(X.T)).dot(y)
```

```python
w_sum = w.sum()
```

```python
print(f"Q7 - Vecteur des poids w : {w}")
print(f"     Somme de tous les éléments de w : {w_sum:.4f}")
```
